```python
def menu():
    # Define a função 'menu'. Nada é executado aqui até chamarmos menu().
    # Funções agrupam código e permitem reutilização/organização.
    while True:
        # 'while True' cria um loop infinito.
        # Esse loop manterá o menu sendo mostrado repetidamente até encontrarmos 'break'.
        print("\n===== SISTEMA DE BIBLIOTECA =====")
        # print envia texto para a saída padrão (terminal).
        # '\n' no começo insere uma linha em branco antes do título para separar visualmente.

        print("1 - Cadastrar livro")
        # Mostra a opção 1 no menu.
        print("2 - Listar livros")
        # Mostra a opção 2 no menu.
        print("3 - Buscar livro")
        # Mostra a opção 3 no menu.
        print("4 - Remover livro")
        # Mostra a opção 4 no menu.
        print("5 - Sair")
        # Mostra a opção 5 no menu.

        opcao = input("Escolha uma opção: ")
        # input exibe o texto do prompt e espera o usuário digitar algo e pressionar Enter.
        # O retorno de input é sempre uma string (por exemplo, '1', '2', 'sair').
        # Aqui armazenamos essa string na variável 'opcao'.

        if opcao == "1":
            # 'if' testa condição de igualdade. Comparação é com string "1"
            # porque input retorna string (não número).
            print("Cadastrar livro ainda não implementado.")
            # Placeholder: mensagem temporária indicando que a função ainda não existe.

        elif opcao == "2":
            # 'elif' = "else if": só é avaliado se o 'if' anterior for falso.
            # Testa se o usuário digitou exatamente "2".
            print("Listar livros ainda não implementado.")
            # Placeholder para futura chamada da função que lista livros.

        elif opcao == "3":
            # Outra verificação: se a opção for "3".
            print("Buscar livro ainda não implementado.")
            # Placeholder para futura função de busca.

        elif opcao == "4":
            # Verifica se a opção é "4".
            print("Remover livro ainda não implementado.")
            # Placeholder para futura função de remoção.

        elif opcao == "5":
            # Verifica se a opção é "5" (sair).
            print("Saindo do sistema...")
            # Mensagem de confirmação.
            break
            # 'break' interrompe o loop 'while True' imediatamente.
            # Isso faz com que a função 'menu' termine (retorne None implicitamente)
            # e o programa continue após a chamada menu() (geralmente termina).

        else:
            # 'else' captura qualquer entrada que não bateu com os 'if' / 'elif' acima.
            print("Opção inválida! Tente novamente.")
            # Informa ao usuário que a entrada não foi reconhecida.
            # Após isso, o loop recomeça e o menu é exibido novamente.

# Chamada da função menu para iniciar o programa.
menu()
# Sem essa linha, a função apenas estaria definida; não seria executada.
```
