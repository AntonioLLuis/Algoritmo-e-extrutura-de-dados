```python
# biblioteca_comentado_linha_a_linha.py
# Versão do sistema de biblioteca com comentários explicativos em cada linha.

# importa o módulo json para ler/escrever dados no formato JSON
import json
# importa o módulo os para verificar existência de arquivos e outras operações do SO
import os

# define uma constante com o nome do arquivo onde os dados serão guardados
ARQUIVO = "biblioteca.json"


# -------------------------------------------
# FUNÇÃO PARA CARREGAR OS LIVROS DO ARQUIVO
# -------------------------------------------
# declara a função carregar_livros que não recebe parâmetros
def carregar_livros():
    # verifica se o arquivo ARQUIVO existe; os.path.exists retorna True/False
    if not os.path.exists(ARQUIVO):
        # se o arquivo não existe, retorna imediatamente uma lista vazia
        return []  # Se não existe o arquivo, retorna lista vazia

    # abre o arquivo no modo leitura ("r") usando um contexto with para garantir fechamento
    with open(ARQUIVO, "r", encoding="utf-8") as f:
        # usa json.load para converter o conteúdo JSON do arquivo numa estrutura Python (lista/dicts)
        return json.load(f)


# -------------------------------------------
# FUNÇÃO PARA SALVAR LIVROS NO ARQUIVO
# -------------------------------------------
# declara a função salvar_livros que recebe a lista 'livros' como parâmetro
def salvar_livros(livros):
    # abre o arquivo no modo escrita ("w"); se não existir, será criado; utf-8 para acentos
    with open(ARQUIVO, "w", encoding="utf-8") as f:
        # json.dump converte a estrutura Python para JSON e grava no arquivo; indent=4 formata legivelmente
        # ensure_ascii=False permite caracteres Unicode (acentos) no arquivo sem escape
        json.dump(livros, f, indent=4, ensure_ascii=False)


# -------------------------------------------
# FUNÇÃO CADASTRAR LIVRO
# -------------------------------------------
# declara a função cadastrar_livro que recebe a lista 'livros' (mutável) para adicionar um novo item
def cadastrar_livro(livros):
    # imprime cabeçalho informativo
    print("\n--- CADASTRAR NOVO LIVRO ---")
    # solicita ao usuário o código do livro; input retorna string; .strip() remove espaços iniciais/finais
    codigo = input("Código do livro: ").strip()

    # inicia um loop for para verificar se o código já existe na lista
    for livro in livros:
        # compara o código informado com o valor do campo "codigo" do dicionário livro
        if livro["codigo"] == codigo:
            # se encontrou código igual, informa ao usuário
            print("Já existe um livro com esse código.")
            # sai da função sem cadastrar (return encerra a execução da função)
            return

    # solicita o título do livro; .strip() remove espaços indesejados
    titulo = input("Título do livro: ").strip()
    # solicita o autor do livro
    autor = input("Autor: ").strip()
    # solicita o ano de publicação
    ano = input("Ano de publicação: ").strip()

    # cria um dicionário que representa o livro com seus campos
    livro = {
        "codigo": codigo,       # armazena o código informado
        "titulo": titulo,       # armazena o título informado
        "autor": autor,         # armazena o autor informado
        "ano": ano,             # armazena o ano informado
        "status": "disponível"  # define status inicial como "disponível"
    }

    # adiciona o dicionário 'livro' ao final da lista 'livros'
    livros.append(livro)
    # chama salvar_livros para persistir imediatamente a alteração no arquivo JSON
    salvar_livros(livros)
    # informa ao usuário que o cadastro foi realizado
    print("Livro cadastrado com sucesso!")


# -------------------------------------------
# FUNÇÃO LISTAR LIVROS
# -------------------------------------------
# declara a função listar_livros que recebe a lista 'livros' para exibir seu conteúdo
def listar_livros(livros):
    # imprime cabeçalho da lista
    print("\n--- LISTA DE LIVROS ---")
    # testa se a lista está vazia; lista vazia é avaliada como False em contexto booleano
    if not livros:
        # avisa que não há livros cadastrados
        print("Nenhum livro cadastrado.")
        # retorna sem continuar (evita tentar iterar lista vazia)
        return

    # itera sobre cada dicionário 'livro' na lista 'livros'
    for livro in livros:
        # usa f-string e string multilinha para imprimir os campos do livro de forma legível
        print(f"""
Código: {livro['codigo']}
Título: {livro['titulo']}
Autor: {livro['autor']}
Ano: {livro['ano']}
Status: {livro['status']}
""")


# -------------------------------------------
# FUNÇÃO BUSCAR LIVRO
# -------------------------------------------
# declara a função buscar_livro que recebe a lista 'livros'
def buscar_livro(livros):
    # imprime cabeçalho da busca
    print("\n--- BUSCAR LIVRO ---")
    # solicita termo de busca; .strip() remove espaços; .lower() normaliza para minúsculas para busca case-insensitive
    termo = input("Digite o título ou código: ").strip().lower()

    # cria lista vazia para guardar os resultados encontrados
    encontrados = []

    # percorre todos os livros procurando correspondência
    for livro in livros:
        # verifica se 'termo' é substring do título em minúsculas ou se é igual ao código
        if termo in livro["titulo"].lower() or termo == livro["codigo"]:
            # se condição verdadeira, adiciona o dicionário 'livro' à lista de encontrados
            encontrados.append(livro)

    # se nenhum livro foi encontrado (lista vazia), informa e retorna
    if not encontrados:
        print("Nenhum livro encontrado.")
        return

    # se encontrou, imprime quantidade de resultados usando len(encontrados)
    print(f"\n{len(encontrados)} livro(s) encontrado(s):\n")
    # itera sobre os livros encontrados para exibir um resumo (código, título, status)
    for livro in encontrados:
        print(f"Código: {livro['codigo']} - {livro['titulo']} ({livro['status']})")


# -------------------------------------------
# FUNÇÃO REMOVER LIVRO
# -------------------------------------------
# declara a função remover_livro que recebe a lista 'livros'
def remover_livro(livros):
    # imprime cabeçalho
    print("\n--- REMOVER LIVRO ---")
    # solicita o código do livro a remover e limpa espaços
    codigo = input("Código do livro a remover: ").strip()

    # percorre a lista para encontrar o livro com o código informado
    for livro in livros:
        # compara o campo "codigo" do dicionário com o código informado
        if livro["codigo"] == codigo:
            # remove o dicionário 'livro' da lista 'livros'
            livros.remove(livro)
            # atualiza o arquivo JSON com a lista modificada
            salvar_livros(livros)
            # confirma a remoção ao usuário
            print("Livro removido com sucesso!")
            # sai da função para não continuar o loop
            return

    # se o loop finalizou sem encontrar o código, informa ao usuário
    print("Nenhum livro encontrado com esse código.")


# -------------------------------------------
# FUNÇÃO EMPRESTAR LIVRO
# -------------------------------------------
# declara a função emprestar_livro que recebe a lista 'livros'
def emprestar_livro(livros):
    # imprime cabeçalho
    print("\n--- EMPRESTAR LIVRO ---")
    # solicita código do livro a ser emprestado e limpa espaços
    codigo = input("Código do livro: ").strip()

    # percorre a lista em busca do livro com o código informado
    for livro in livros:
        # compara o código do livro atual com o informado
        if livro["codigo"] == codigo:
            # se o campo status for 'emprestado', não é possível emprestar novamente
            if livro["status"] == "emprestado":
                # informa usuário que já está emprestado
                print("Este livro já está emprestado!")
                # sai da função
                return

            # caso esteja disponível, altera o status para 'emprestado'
            livro["status"] = "emprestado"
            # salva as modificações no arquivo JSON
            salvar_livros(livros)
            # confirma o empréstimo ao usuário
            print("Livro emprestado com sucesso!")
            # sai da função após registrar o empréstimo
            return

    # se não encontrou o livro (nenhum if bateu), informa que não foi encontrado
    print("Livro não encontrado.")


# -------------------------------------------
# FUNÇÃO DEVOLVER LIVRO
# -------------------------------------------
# declara a função devolver_livro que recebe a lista 'livros'
def devolver_livro(livros):
    # imprime cabeçalho
    print("\n--- DEVOLVER LIVRO ---")
    # solicita o código do livro a ser devolvido e limpa espaços
    codigo = input("Código do livro: ").strip()

    # percorre os livros para encontrar o código informado
    for livro in livros:
        # compara o código do dicionário com o código informado
        if livro["codigo"] == codigo:
            # se o status já for 'disponível', não há devolução necessária
            if livro["status"] == "disponível":
                # informa ao usuário
                print("Este livro já está disponível!")
                # sai da função
                return

            # caso esteja emprestado, altera o status para 'disponível'
            livro["status"] = "disponível"
            # salva a alteração no arquivo JSON
            salvar_livros(livros)
            # confirma a devolução
            print("Livro devolvido com sucesso!")
            # sai da função após processar a devolução
            return

    # se não encontrou o livro com o código informado, informa o usuário
    print("Livro não encontrado.")


# -------------------------------------------
# MENU PRINCIPAL
# -------------------------------------------
# declara a função menu que orquestra todo o sistema
def menu():
    # carrega a lista de livros do arquivo chamando carregar_livros()
    livros = carregar_livros()

    # inicia o loop principal do menu; repetirá até o usuário escolher sair
    while True:
        # imprime um cabeçalho do sistema com uma linha em branco no início
        print("\n===== SISTEMA DE BIBLIOTECA =====")
        # exibe as opções numeradas do menu
        print("1 - Cadastrar livro")
        print("2 - Listar livros")
        print("3 - Buscar livro")
        print("4 - Remover livro")
        print("5 - Emprestar livro")
        print("6 - Devolver livro")
        print("7 - Sair")

        # lê a opção do usuário como string (input sempre retorna string)
        opcao = input("Escolha uma opção: ")

        # usa estrutura condicional para chamar a função correspondente à opção
        if opcao == "1":
            # chama a função cadastrar_livro passando a lista 'livros' (por referência)
            cadastrar_livro(livros)
        elif opcao == "2":
            # chama a função listar_livros para exibir todos os registros
            listar_livros(livros)
        elif opcao == "3":
            # chama a função buscar_livro para procurar por título ou código
            buscar_livro(livros)
        elif opcao == "4":
            # chama a função remover_livro que exclui um item por código
            remover_livro(livros)
        elif opcao == "5":
            # chama a função emprestar_livro para marcar um livro como emprestado
            emprestar_livro(livros)
        elif opcao == "6":
            # chama a função devolver_livro para registrar a devolução
            devolver_livro(livros)
        elif opcao == "7":
            # informa que o sistema vai encerrar
            print("Saindo do sistema...")
            # break encerra o loop while, fazendo com que a função menu termine
            break
        else:
            # caso a entrada não corresponda a nenhuma opção válida, informa o usuário
            print("Opção inválida! Tente novamente.")


# chamada que inicia o programa ao executar o arquivo
menu()
```
