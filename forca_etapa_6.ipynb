Etapa 6 — Refatorando para funções (melhor organização)

Vamos separar responsabilidades em funções para o código ficar limpo e fácil de manter.

```python
# forca_funcoes_etapa6.py
import random
import os

# Lista de palavras possíveis (sem acentos)
WORD_LIST = ["python", "computador", "algoritmo", "projeto", "arquivo"]

# Desenhos ASCII da forca para cada número de erros (do 0 ao máximo)
# Cada índice representa o estado da forca com aquele número de erros.
FORCA_STAGES = [
    """
     _______
    |/      |
    |
    |
    |
    |
    |
    |___
    """,
    """
     _______
    |/      |
    |      (_)
    |
    |
    |
    |
    |___
    """,
    """
     _______
    |/      |
    |      (_)
    |       |
    |       |
    |
    |
    |___
    """,
    """
     _______
    |/      |
    |      (_)
    |      \\|
    |       |
    |
    |
    |___
    """,
    """
     _______
    |/      |
    |      (_)
    |      \\|/
    |       |
    |
    |
    |___
    """,
    """
     _______
    |/      |
    |      (_)
    |      \\|/
    |       |
    |      /
    |
    |___
    """,
    """
     _______
    |/      |
    |      (_)
    |      \\|/
    |       |
    |      / \\
    |
    |___
    """,
]

def limpar_tela():
    """Tenta limpar a tela do terminal para melhor visualização (Windows/Linux/Mac)."""
    # função simples e segura: tenta os dois comandos
    os.system("cls" if os.name == "nt" else "clear")

def escolher_palavra(palavras):
    """Retorna uma palavra aleatória da lista fornecida."""
    return random.choice(palavras)

def desenhar_forca(erros, max_erros):
    """
    Exibe o desenho da forca correspondente ao número de erros.
    Se max_erros for maior que o número de imagens disponíveis,
    normaliza para o último estágio.
    """
    idx = min(erros, len(FORCA_STAGES) - 1)  # garante índice válido
    print(FORCA_STAGES[idx])
    # também mostramos progresso numérico
    print(f"Erros: {erros}/{max_erros}\n")

def mostrar_estado(palavra, letras_reveladas, erros, max_erros, letras_chutadas):
    """Mostra o estado atual do jogo: forca, palavra parcial e letras chutadas."""
    desenhar_forca(erros, max_erros)
    print("Palavra:", " ".join(letras_reveladas))
    print("Letras chutadas:", " ".join(sorted(letras_chutadas)))
    print()

def selecionar_dificuldade():
    """Permite ao jogador escolher a dificuldade e retorna max_erros."""
    print("Escolha a dificuldade:")
    print("1 - Fácil (8 erros)")
    print("2 - Médio (6 erros)")
    print("3 - Difícil (4 erros)")
    escolha = input("Digite 1, 2 ou 3 (ENTER para Médio): ").strip()
    if escolha == "1":
        return 8
    if escolha == "3":
        return 4
    # padrão: médio
    return 6

def jogar_uma_partida(palavras, max_erros=6, limpar=True):
    """
    Executa uma partida de forca.
    Retorna True se o jogador venceu, False caso perdeu, None se desistiu.
    """
    palavra = escolher_palavra(palavras)
    letras_reveladas = ["_"] * len(palavra)
    erros = 0
    letras_chutadas = set()

    while True:
        if limpar:
            limpar_tela()
        mostrar_estado(palavra, letras_reveladas, erros, max_erros, letras_chutadas)
        chute = input("Letra / palavra / desistir: ").strip().lower()
        if not chute:
            print("Digite algo.")
            input("Enter para continuar...")
            continue

        if chute == "desistir":
            print("Você desistiu. Palavra:", palavra)
            return None

        # tentativa de palavra inteira
        if len(chute) > 1:
            if chute == palavra:
                print("Você acertou! Palavra:", palavra)
                return True
            else:
                erros += 1
                print("Palavra incorreta.")
                if erros < max_erros:
                    input("Enter para continuar...")
        else:
            letra = chute[0]
            if letra in letras_chutadas:
                print("Você já chutou essa letra.")
                input("Enter para continuar...")
                continue
            letras_chutadas.add(letra)
            if letra in palavra:
                for i, ch in enumerate(palavra):
                    if ch == letra:
                        letras_reveladas[i] = ch
            else:
                erros += 1
                print("Letra errada.")
                if erros < max_erros:
                    input("Enter para continuar...")

        # Vitória por revelar todas as letras
        if "_" not in letras_reveladas:
            if limpar:
                limpar_tela()
            mostrar_estado(palavra, letras_reveladas, erros, max_erros, letras_chutadas)
            print("Você venceu! Palavra:", palavra)
            return True

        # Derrota por atingir o máximo de erros
        if erros >= max_erros:
            if limpar:
                limpar_tela()
            desenhar_forca(erros, max_erros)
            print("Você perdeu! Palavra era:", palavra)
            return False

def main():
    """Loop principal: gerencia placar e permite jogar novamente."""
    limpar_tela()
    print("=== Jogo da Forca — Etapa 6 ===")
    max_erros = selecionar_dificuldade()
    wins = 0
    losses = 0
    desistencias = 0

    while True:
        resultado = jogar_uma_partida(WORD_LIST, max_erros=max_erros, limpar=True)
        if resultado is True:
            wins += 1
        elif resultado is False:
            losses += 1
        else:
            desistencias += 1

        # mostra placar
        print("\nPlacar atual:")
        print(f"Vitórias: {wins}  Derrotas: {losses}  Desistências: {desistencias}\n")

        # pergunta se quer jogar novamente
        again = input("Jogar novamente? (s/n) ").strip().lower()
        if again != "s":
            print("Obrigado por jogar! Até a próxima.")
            break

if __name__ == "__main__":
    main()
```
