Etapa 6 — Refatorando para funções (melhor organização)

Vamos separar responsabilidades em funções para o código ficar limpo e fácil de manter.

```python
# forca_funcoes.py

# Importa o módulo random para permitir sorteio aleatório de palavras
import random

# Lista de palavras possíveis para o jogo (sem acentos)
WORD_LIST = ["python", "computador", "algoritmo", "projeto", "arquivo"]

# Função que escolhe e devolve uma palavra aleatória a partir da lista passada
def escolher_palavra(palavras):
    return random.choice(palavras)

# Função que exibe o estado atual do jogo: palavra parcial, erros e letras chutadas
def mostrar_estado(palavra, letras_reveladas, erros, max_erros, letras_chutadas):
    # Mostra a palavra com espaços entre os caracteres (ex.: _ _ t h o n)
    print("Palavra:", " ".join(letras_reveladas))
    # Mostra o número de erros atual e o máximo permitido
    print(f"Erros: {erros}/{max_erros}")
    # Mostra as letras já chutadas (ordenadas), separadas por espaço
    print("Letras chutadas:", " ".join(sorted(letras_chutadas)))
    # Linha em branco para separar visualmente as informações
    print()

# Função principal que executa o jogo; recebe lista de palavras e max_erros opcional
def jogar(palavras, max_erros=6):
    # Sorteia uma palavra usando a função escolher_palavra
    palavra = escolher_palavra(palavras)
    # Inicializa a representação das letras reveladas com "_" para cada letra da palavra
    letras_reveladas = ["_"] * len(palavra)
    # Inicializa o contador de erros
    erros = 0
    # Conjunto para registrar letras chutadas (evita duplicatas)
    letras_chutadas = set()

    # Loop principal do jogo — repete até vitória, derrota ou desistência
    while True:
        # Mostra o estado atual do jogo chamando mostrar_estado
        mostrar_estado(palavra, letras_reveladas, erros, max_erros, letras_chutadas)
        # Lê o chute do usuário; pode ser letra, tentativa de palavra ou "desistir"
        chute = input("Letra / palavra / desistir: ").strip().lower()
        # Se o usuário não digitou nada (string vazia), pede para digitar algo e reinicia o loop
        if not chute:
            print("Digite algo.")
            continue
        # Se o usuário digitou exatamente 'desistir', informa a palavra e encerra o jogo
        if chute == "desistir":
            print("Você desistiu. Palavra:", palavra)
            break
        # Se o usuário digitou mais de um caractere, interpretamos como tentativa de palavra inteira
        if len(chute) > 1:
            # Se a tentativa for exatamente a palavra sorteada, o jogador vence
            if chute == palavra:
                print("Você acertou! Palavra:", palavra)
                break
            else:
                # Tentativa incorreta de palavra inteira conta como erro
                erros += 1
                print("Palavra incorreta.")
        else:
            # Caso normal: o usuário chutou uma única letra
            letra = chute[0]
            # Se essa letra já foi chutada antes, informa e volta ao início do loop (sem penalizar)
            if letra in letras_chutadas:
                print("Você já chutou essa letra.")
                continue
            # Adiciona a letra ao conjunto de letras chutadas
            letras_chutadas.add(letra)
            # Se a letra está presente na palavra, revela todas as ocorrências
            if letra in palavra:
                for i, ch in enumerate(palavra):
                    if ch == letra:
                        letras_reveladas[i] = ch
            else:
                # Se a letra não existe na palavra, incrementa o contador de erros
                erros += 1
                print("Letra errada.")
        # Verifica se todas as letras foram reveladas (vitória)
        if "_" not in letras_reveladas:
            print("Você venceu! Palavra:", palavra)
            break
        # Verifica se o jogador atingiu ou ultrapassou o número máximo de erros (derrota)
        if erros >= max_erros:
            print("Você perdeu! Palavra era:", palavra)
            break

# Ponto de entrada: só executa jogar() quando o arquivo é executado diretamente
if __name__ == "__main__":
    jogar(WORD_LIST, max_erros=6)
```
