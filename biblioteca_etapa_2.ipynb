```python
livros = []
# Criamos uma lista vazia chamada 'livros'.
# Essa lista servirá para armazenar vários dicionários.
# Cada dicionário representará um livro com seus dados:
# código, título, autor e status.


def menu():
    # Início da função 'menu', que controla toda a interface do usuário.

    while True:
        # Loop infinito para manter o menu sempre aparecendo até o usuário escolher sair.

        print("\n1 - Cadastrar")
        # '\n' adiciona uma linha em branco para melhorar a visualização.
        # Exibe a opção de cadastrar um livro.

        print("2 - Listar")
        # Exibe a opção de listar todos os livros cadastrados.

        print("3 - Sair")
        # Exibe a opção de encerrar o programa.

        opcao = input("Opção: ")
        # Solicita ao usuário que escolha uma das opções acima.
        # O input sempre retorna texto (string).

        if opcao == "1":
            # Se a opção digitada for exatamente "1", entramos no processo de cadastro.

            # A seguir, pedimos os dados do livro. Tudo ainda bem simples nesta etapa.

            codigo = input("Código: ").strip()
            # input() lê o texto digitado.
            # .strip() remove espaços extras do início e do final (útil para evitar erros).

            titulo = input("Título: ").strip()
            # Lê o título do livro e remove espaços.

            autor = input("Autor: ").strip()
            # Lê o autor do livro e remove espaços.

            livro = {"codigo": codigo, "titulo": titulo, "autor": autor, "status": "disponível"}
            # Criamos um dicionário representando um livro.
            # Um dicionário é uma estrutura chave-valor, por exemplo:
            # {"codigo": "001", "titulo": "Python Básico", "autor": "João", "status": "disponível"}
            # O campo 'status' começará sempre como "disponível".

            livros.append(livro)
            # .append() adiciona o dicionário 'livro' ao final da lista 'livros'.
            # Agora a lista tem 1 livro cadastrado (ou mais, se repetirmos a operação).

        elif opcao == "2":
            # Se a opção for "2", queremos listar todos os livros cadastrados.

            for l in livros:
                # 'for' percorre cada elemento da lista 'livros'.
                # Cada elemento da lista é um dicionário → representado aqui pela variável 'l'.

                print(l)
                # Imprime o dicionário inteiro.
                # Exemplo de saída: {'codigo': '001', 'titulo': 'Python', 'autor': 'Ana', 'status': 'disponível'}

        elif opcao == "3":
            # Se a opção for "3", o usuário quer sair do sistema.

            break
            # O comando 'break' encerra o loop 'while True'.
            # Isso faz com que o programa saia da função menu() e termine.


# Chamada da função menu para iniciar a execução do sistema.
menu()
# Se essa linha não existisse, nada seria executado.
```
