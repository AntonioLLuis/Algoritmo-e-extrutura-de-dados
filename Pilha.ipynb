```
# Importa a biblioteca NumPy e a renomeia como np.
# NumPy é usada aqui para criar um array de forma eficiente para armazenar os valores da pilha.
import numpy as np

# Define a classe Pilha.
# Uma Pilha é uma estrutura de dados Linear do tipo LIFO (Last-In, First-Out).
class Pilha:

  # Método construtor da classe. Ele é chamado quando um novo objeto Pilha é criado.
  # 'capacidade' é o tamanho máximo que a pilha pode ter.
  def __init__(self, capacidade):
    # Atributo privado que armazena a capacidade máxima da pilha (tamanho fixo).
    # O uso de '__' indica que o atributo é "privado" (convenção em Python para evitar acesso direto).
    self.__capacidade = capacidade

    # Atributo privado que armazena a posição do elemento do topo da pilha.
    # Inicialmente, a pilha está vazia, então o topo é setado para -1.
    self.__topo = -1

    # Atributo privado que armazena os valores da pilha em um array NumPy.
    # np.empty(self.__capacidade, dtype=int) cria um array com o tamanho da capacidade
    # e especifica que os elementos serão inteiros (dtype=int).
    # Os valores iniciais do array são lixo de memória e não são importantes, pois o controle é feito por __topo.
    self.__valores = np.empty(self.__capacidade, dtype=int)

  # Método privado para verificar se a pilha está cheia.
  # Geralmente usado internamente pela classe antes de empilhar.
  def __pilha_cheia(self):
    # Compara se a posição do topo é igual ao índice final do array (capacidade - 1).
    if self.__topo == self.__capacidade - 1:
      # Se for igual, a pilha está cheia.
      return True
    else:
      # Caso contrário, a pilha não está cheia.
      return False

  # Método privado para verificar se a pilha está vazia.
  # Geralmente usado internamente pela classe antes de desempilhar ou ver o topo.
  def __pilha_vazia(self):
    # Compara se a posição do topo é igual a -1.
    if self.__topo == -1:
      # Se for -1, a pilha está vazia.
      return True
    else:
      # Caso contrário, a pilha contém elementos.
      return False

  # Método público para adicionar um elemento (valor) ao topo da pilha (operação PUSH).
  def empilhar(self, valor):
    # Verifica se a pilha está cheia usando o método auxiliar privado.
    if self.__pilha_cheia():
      # Se estiver cheia, exibe uma mensagem de erro e não faz nada.
      print('A pilha está cheia')
    else:
      # Se não estiver cheia:
      # 1. Incrementa o ponteiro do topo para a próxima posição livre.
      self.__topo += 1
      # 2. Insere o 'valor' na nova posição do topo.
      self.__valores[self.__topo] = valor
      #  # Opcional: Diagrama de empilhamento

  # Método público para remover o elemento do topo da pilha (operação POP).
  # Em estruturas de Pilha, o elemento removido é geralmente o último que foi adicionado (LIFO).
  def desempilhar(self):
    # Verifica se a pilha está vazia usando o método auxiliar privado.
    if self.__pilha_vazia():
      # Se estiver vazia, exibe uma mensagem de erro e não faz nada.
      print('A pilha está vazia')
    else:
      # Se não estiver vazia:
      # 1. Decrementa o ponteiro do topo. O valor anterior ainda está no array,
      #    mas ele não é mais considerado parte da pilha.
      self.__topo -= 1
      # Nota: Em Python, o valor removido (desempilhado) é tipicamente retornado.
      # Esta implementação apenas remove logicamente o elemento, sem retorná-lo.
      #  # Opcional: Diagrama de desempilhamento

  # Método público para retornar o valor do topo da pilha sem removê-lo (operação PEEK/TOP).
  def ver_topo(self):
    # Verifica se a pilha não está vazia.
    if self.__topo != -1:
      # Se não estiver vazia, retorna o valor que está na posição atual do topo.
      return self.__valores[self.__topo]
    else:
      # Se estiver vazia, retorna -1 (ou outro valor que indique que não há topo para ver).
      return -1
```

```
pilha = Pilha(5)
```

```
pilha.ver_topo()
```

```
pilha.empilhar(1)
pilha.ver_topo()
```

```
pilha.empilhar(1)
pilha.empilhar(2)
pilha.empilhar(3)
pilha.empilhar(4)
```

```
pilha.empilhar(6)
```

```
pilha.ver_topo()
```

```
pilha.desempilhar()
```

```
pilha.ver_topo()
```

```
pilha.desempilhar()
pilha.desempilhar()
pilha.desempilhar()
```
