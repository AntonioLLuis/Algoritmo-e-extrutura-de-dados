Etapa 5 — Tentar adivinhar a palavra inteira e opção "desistir"

Permitir que o jogador digite mais de uma letra para tentar a palavra inteira. Também adicionamos a palavra-chave desistir.

Explicação das adições

Se chute == "desistir", terminamos o jogo mostrando a palavra.

Se len(chute) > 1, interpretamos como tentativa de adivinhar a palavra inteira (acerto = vitória; erro = +1 erro).

Mantemos a lógica anterior para chutes de letra.

Teste: tente adivinhar a palavra inteira; veja como conta como erro quando errada.

```python
# Importa o módulo random para selecionar uma palavra aleatoriamente
import random

# Lista de palavras possíveis (sem acentos)
WORD_LIST = ["python", "computador", "algoritmo", "projeto", "arquivo"]

# Sorteia uma palavra da lista
palavra = random.choice(WORD_LIST)

# Cria a representação das letras reveladas com "_" para cada letra da palavra
letras_reveladas = ["_"] * len(palavra)

# Inicializa o contador de erros
erros = 0

# Define o máximo de erros permitidos antes de perder
MAX_ERROS = 6

# Conjunto para registrar letras já chutadas (evita duplicatas)
letras_chutadas = set()

# Loop principal do jogo: continua até vitória, derrota ou desistência
while True:
    # Mostra a palavra com as letras descobertas e ocultas
    print("Palavra:", " ".join(letras_reveladas))
    # Mostra o contador de erros atual e o limite
    print(f"Erros: {erros}/{MAX_ERROS}")
    # Mostra as letras já tentadas em ordem alfabética
    print("Letras chutadas:", " ".join(sorted(letras_chutadas)))
    # Lê o chute do jogador — pode ser: uma letra, uma palavra inteira ou 'desistir'
    chute = input("Digite uma letra, tente a palavra inteira ou 'desistir': ").strip().lower()

    # Se não digitou nada, pede de novo
    if not chute:
        print("Digite algo.")
        continue  # volta ao início do while

    # Se o jogador digitou exatamente 'desistir', encerra o jogo informando a palavra
    if chute == "desistir":
        print("Você desistiu. Palavra:", palavra)
        break  # termina o loop e o jogo

    # Se o jogador digitou mais de um caractere, consideramos uma tentativa de adivinhar a palavra inteira
    if len(chute) > 1:
        # Se a tentativa for exatamente a palavra sorteada, o jogador vence
        if chute == palavra:
            print("Você acertou! Palavra:", palavra)
            break  # encerra o jogo com vitória
        else:
            # Tentativa de palavra incorreta: contabiliza um erro
            erros += 1
            print("Palavra incorreta.")
    else:
        # Caso normal: o jogador chutou uma única letra
        letra = chute[0]  # pega o primeiro (e único) caractere
        # Se a letra já foi chutada antes, avisa e volta ao loop (não conta erro)
        if letra in letras_chutadas:
            print("Você já chutou essa letra.")
            continue
        # Registra a letra como chutada
        letras_chutadas.add(letra)
        # Se a letra está na palavra, revela todas as ocorrências
        if letra in palavra:
            for i, ch in enumerate(palavra):
                if ch == letra:
                    letras_reveladas[i] = ch
        else:
            # Letra não está na palavra: incrementa erros e avisa
            erros += 1
            print("Letra errada.")

    # Verifica se não existem mais "_" — ou seja, a palavra foi toda descoberta
    if "_" not in letras_reveladas:
        print("Você venceu! Palavra:", palavra)
        break  # encerra com vitória

    # Verifica se o jogador atingiu o limite de erros — se sim, ele perde
    if erros >= MAX_ERROS:
        print("Você perdeu! Palavra era:", palavra)
        break  # encerra com derrota
```
