```python
def pedir_nome():
    nome = input("Digite o nome do aluno: ")
    return nome
```

```python
def pedir_notas():
    notas = []
    for i in range(1, 5):
        nota = float(input(f"Digite a nota do {i}º bimestre: "))
        notas.append(nota)
    return notas
```

```python
def somar_notas(notas):
    total = sum(notas)
    return total
```

```python
def verificar_aprovacao(total):
    if total >= 60:
        return "Aprovado"
    else:
        return "Reprovado"
```

```python
def imprimir_boletim(nome, notas, total, situacao):
    print("\n=== BOLETIM ESCOLAR ===")
    print(f"Aluno: {nome}")
    for i, nota in enumerate(notas, 1):
        print(f"Nota {i}º bimestre: {nota}")
    print(f"Total de pontos: {total}")
    print(f"Situação: {situacao}")
```

```python
nome = pedir_nome()
notas = pedir_notas()
total = somar_notas(notas)
situacao = verificar_aprovacao(total)
imprimir_boletim(nome, notas, total, situacao)
```
