Resumo das novidades da Etapa 4
1. Armazenar letras já chutadas
letras_chutadas = set()


Usa-se um set porque:

não permite duplicatas

é rápido para verificar if letra in letras_chutadas

sorted() organiza em ordem alfabética

join() transforma em string com espaços

    continue


Impede penalizar o jogador repetidas vezes.

4. Registrar o chute
letras_chutadas.add(letra)

```python
# Importa o módulo random para escolher uma palavra aleatória
import random

# Lista com palavras possíveis para o jogo (todas sem acentos)
WORD_LIST = ["python", "computador", "algoritmo", "projeto", "arquivo"]

# Escolhe aleatoriamente uma das palavras da lista
palavra = random.choice(WORD_LIST)

# Lista que mostrará as letras descobertas, iniciada com "_" para cada letra
letras_reveladas = ["_"] * len(palavra)

# Contador de erros do jogador
erros = 0

# Número máximo de erros permitidos antes de perder o jogo
MAX_ERROS = 6

# Conjunto (set) para guardar letras já chutadas, evitando repetição
letras_chutadas = set()

# Loop principal do jogo (continua até vitória ou derrota)
while True:
    # Mostra a palavra atual com espaços entre as letras reveladas/ocultas
    print("Palavra:", " ".join(letras_reveladas))

    # Mostra quantos erros o jogador tem até agora
    print(f"Erros: {erros}/{MAX_ERROS}")

    # Mostra todas as letras já chutadas, em ordem alfabética (sorted)
    print("Letras chutadas:", " ".join(sorted(letras_chutadas)))

    # Recebe o chute do usuário, remove espaços e transforma em minúsculo
    chute = input("Digite uma letra: ").strip().lower()

    # Caso o usuário não digite nada (pressionar ENTER)
    if not chute:
        print("Digite algo.")
        continue  # volta para pedir novamente

    # Considera apenas o primeiro caractere digitado
    letra = chute[0]

    # Verifica se a letra já foi chutada antes
    if letra in letras_chutadas:
        print("Você já chutou essa letra.")
        continue  # não avança o jogo, apenas ignora o chute repetido

    # Adiciona a letra ao conjunto de letras chutadas
    letras_chutadas.add(letra)

    # Se a letra está dentro da palavra
    if letra in palavra:
        # Revela todas as ocorrências dessa letra dentro da palavra
        for i, ch in enumerate(palavra):
            if ch == letra:
                letras_reveladas[i] = ch
    else:
        # Se a letra não faz parte da palavra, soma um erro
        erros += 1
        print("Letra errada.")

    # Se não houver mais "_" na lista, significa que o jogador terminou a palavra
    if "_" not in letras_reveladas:
        print("Você venceu! Palavra:", palavra)
        break  # encerra o loop

    # Se o número de erros atingir o máximo permitido
    if erros >= MAX_ERROS:
        print("Você perdeu! Palavra era:", palavra)
        break  # encerra o loop
```
