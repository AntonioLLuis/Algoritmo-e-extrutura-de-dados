```python
# BOLETIM ESCOLAR
def pedir_nome():
    while True:
        nome = input("Digite o nome do aluno: ").strip()  # .strip() remove espaços antes/depois
        if nome == "":  # Se o aluno não digitar nada
            print("❌ O nome não pode estar vazio. Tente novamente.")
        else:
            return nome  # Retorna o nome digitado e encerra a função
```

```python

# -------------------------------------------------------------
# Função que pede as notas dos 4 bimestres (0 a 25)
# -------------------------------------------------------------
def pedir_notas():
    notas = []  # Lista para guardar as 4 notas do aluno

    for i in range(1, 5):  # Repete 4 vezes (1º ao 4º bimestre)
        while True:
            try:
                # Pede a nota e converte para número decimal
                nota = float(input(f"Digite a nota do {i}º bimestre (0 a 25): "))

                # Verifica se a nota está dentro do intervalo permitido
                if 0 <= nota <= 25:
                    notas.append(nota)  # Adiciona a nota na lista
                    break  # Sai do while e passa para o próximo bimestre
                else:
                    print("⚠️  A nota deve estar entre 0 e 25.")
            except ValueError:
                # Caso o aluno digite algo que não seja número
                print("❌ Entrada inválida! Digite apenas números.")
    return notas  # Retorna a lista de notas
```

```python
# -------------------------------------------------------------
# Função que soma todas as notas
# -------------------------------------------------------------
def somar_notas(notas):
    total = sum(notas)  # Soma todos os valores dentro da lista
    return total
```

```python
# -------------------------------------------------------------
# Função que verifica se o aluno foi aprovado, recuperação ou reprovado
# -------------------------------------------------------------
def verificar_aprovacao(total):
    media = total / 4  # Calcula a média (dividindo o total por 4 bimestres)

    # Critérios de aprovação
    if media >= 15:  # Média de 15 (60 pontos de 100)
        return "Aprovado"
    elif 10 <= media < 15:  # Entre 10 e 14,99 (40 a 59 pontos)
        return "Recuperação"
    else:
        return "Reprovado"
```

```python
# -------------------------------------------------------------
# Função que imprime o boletim completo na tela
# -------------------------------------------------------------
def imprimir_boletim(nome, notas, total, situacao):
    print("\n=== BOLETIM ESCOLAR ===")
    print(f"Aluno: {nome}")

    # Mostra as notas de cada bimestre
    for i, nota in enumerate(notas, 1):  # enumerate começa a contagem em 1
        print(f"Nota {i}º bimestre: {nota}")

    print(f"Total de pontos: {total}")
    print(f"Média: {total / 4:.2f}")  # Mostra a média com duas casas decimais
    print(f"Situação: {situacao}")
```

```python
# -------------------------------------------------------------
# Função principal — executa o programa completo
# -------------------------------------------------------------
def main():
    nome = pedir_nome()  # Pede o nome do aluno
    notas = pedir_notas()  # Pede as notas (0 a 25)
    total = somar_notas(notas)  # Soma as notas
    situacao = verificar_aprovacao(total)  # Verifica a situação
    imprimir_boletim(nome, notas, total, situacao)  # Mostra o boletim completo


# -------------------------------------------------------------
# Ponto de entrada do programa
# -------------------------------------------------------------
if __name__ == "__main__":
    main()
```
