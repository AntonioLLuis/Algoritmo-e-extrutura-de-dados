```python
# forca_simples.py
palavra = "python"                     # palavra fixa, sem acento
letras_reveladas = ["_"] * len(palavra)  # cria lista de underscores do mesmo tamanho da palavra

while True:                             # loop principal do jogo (repetição até break)
    print("Palavra:", " ".join(letras_reveladas))   # mostra a palavra com espaços entre os caracteres
    chute = input("Digite uma letra: ").strip().lower()  # lê entrada, remove espaços e converte pra minúsculas
    if not chute:                        # verifica se a string está vazia (usuário só apertou Enter)
        print("Digite algo.")
        continue                         # volta ao início do while sem processar mais
    letra = chute[0]                     # pega o primeiro caractere da entrada
    # revelar ocorrências
    for i, ch in enumerate(palavra):     # percorre a palavra com índices (i) e caracteres (ch)
        if ch == letra:
            letras_reveladas[i] = ch     # substitui o '_' pela letra correspondente na posição i
    # checar vitória
    if "_" not in letras_reveladas:      # se não há '_' na lista, todas as letras foram reveladas
        print("Você venceu! Palavra:", palavra)
        break                            # encerra o loop (jogo termina)
```
