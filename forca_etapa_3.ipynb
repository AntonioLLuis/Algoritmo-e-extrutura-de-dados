```python
# Importa o módulo random para escolher uma palavra aleatoriamente
import random

# Define a lista de palavras possíveis (sem acentos)
WORD_LIST = ["python", "computador", "algoritmo", "projeto", "arquivo"]

# Escolhe aleatoriamente uma palavra da lista WORD_LIST
palavra = random.choice(WORD_LIST)

# Cria uma lista com "_" para cada letra da palavra: representa letras ainda não reveladas
letras_reveladas = ["_"] * len(palavra)

# Inicializa o contador de erros com zero
erros = 0

# Define o número máximo de erros permitidos antes do jogador perder
MAX_ERROS = 6  # você pode ajustar esse valor para ficar mais fácil/difícil

# Inicia um loop infinito; o jogo continuará até ocorrer vitória ou derrota (break)
while True:
    # Mostra a palavra com espaços entre os caracteres (ex.: _ _ t h o n)
    print("Palavra:", " ".join(letras_reveladas))

    # Exibe na tela o número atual de erros e o máximo permitido (ex.: Erros: 2/6)
    print(f"Erros: {erros}/{MAX_ERROS}")

    # Lê o chute do usuário, remove espaços em branco nas extremidades e converte para minúsculas
    chute = input("Digite uma letra: ").strip().lower()

    # Se o usuário não digitou nada (string vazia), avisa e recomeça o loop para pedir novamente
    if not chute:
        print("Digite algo.")
        continue  # volta ao início do while sem executar o resto do bloco

    # Pega apenas o primeiro caractere digitado pelo usuário (caso ele tenha digitado mais de 1)
    letra = chute[0]

    # Verifica se a letra digitada está presente na palavra sorteada
    if letra in palavra:
        # Se a letra está na palavra, percorre cada caractere com índice usando enumerate
        for i, ch in enumerate(palavra):
            # Se a letra da posição atual for igual à letra chutada...
            if ch == letra:
                # ...revela a letra na lista letras_reveladas na mesma posição
                letras_reveladas[i] = ch
    else:
        # Se a letra não estiver na palavra, incrementa o contador de erros
        erros += 1
        # Informa ao jogador que a letra está errada
        print("Letra errada.")

    # Verifica se não existe mais "_" em letras_reveladas (todas as letras foram descobertas)
    if "_" not in letras_reveladas:
        # Se verdade, o jogador venceu — mostra a palavra e encerra o loop
        print("Você venceu! Palavra:", palavra)
        break  # termina o while

    # Verifica se o número de erros chegou ou ultrapassou o máximo permitido
    if erros >= MAX_ERROS:
        # Se verdade, o jogador perdeu — mostra qual era a palavra e encerra o loop
        print("Você perdeu! Palavra era:", palavra)
        break  # termina o while
```
