```python
# Importa o módulo random, que permite gerar números aleatórios
import random

# Cria uma lista com várias palavras (todas sem acento)
WORD_LIST = ["python", "computador", "algoritmo", "projeto", "arquivo"]

# Escolhe aleatoriamente uma palavra da lista
palavra = random.choice(WORD_LIST)

# Cria uma lista com "_" para cada letra da palavra, representando letras ainda não descobertas
letras_reveladas = ["_"] * len(palavra)

# Inicia um loop infinito — o jogo só termina quando o jogador vencer
while True:
    # Mostra na tela a palavra com espaços entre os caracteres (ex: _ _ _ _ _)
    print("Palavra:", " ".join(letras_reveladas))

    # Lê uma letra do usuário, remove espaços e converte para minúscula
    chute = input("Digite uma letra: ").strip().lower()

    # Se o usuário apenas apertou ENTER, avisa e continua o loop
    if not chute:
        print("Digite algo.")
        continue

    # Pega apenas o primeiro caractere digitado (ex: se digitar "ab", pega "a")
    letra = chute[0]

    # Percorre cada posição da palavra
    for i, ch in enumerate(palavra):
        # Se a letra digitada for igual à letra da palavra nessa posição...
        if ch == letra:
            # ...revela a letra na posição correta
            letras_reveladas[i] = ch

    # Se não houver mais "_" na lista, significa que a palavra foi toda descoberta
    if "_" not in letras_reveladas:
        print("Você venceu! Palavra:", palavra)
        break  # encerra o jogo
```
