```python

def pedir_nome_produto():
    # Pede ao usuário para digitar o nome do produto
    nome = input('Digite o nome do produto: ')
    # Retorna o nome digitado para ser usado em outras partes do programa
    return nome


# -------------------------------
# Função que pede o estoque inicial do produto
# -------------------------------
def pedir_estoque_inicial():
    # Converte o valor digitado pelo usuário em número inteiro
    estoque = int(input('Digite a quantidade inicial em estoque: '))
    # Retorna o valor digitado
    return estoque


# -------------------------------
# Função que pede a quantidade de vendas
# -------------------------------
def pedir_vendas():
    # Pede ao usuário quantas unidades foram vendidas e converte para inteiro
    vendas = int(input('Digite quantas unidades foram vendidas: '))
    # Retorna o valor das vendas
    return vendas


# -------------------------------
# Função que calcula o estoque final
# -------------------------------
def calcular_estoque_final(inicial, vendas):
    # Subtrai o número de vendas do estoque inicial
    final = inicial - vendas
    # Retorna o resultado do cálculo (estoque final)
    return final


# -------------------------------
# Função que verifica a situação do estoque
# -------------------------------
def verificar_situacao(estoque_final):
    # Se o estoque for 10 ou mais, ainda há bastante produto
    if estoque_final >= 10:
        return 'Estoque suficiente'
    # Se o estoque estiver entre 1 e 9, está acabando
    elif estoque_final >= 1:
        return 'Estoque baixo'
    # Se o estoque for 0 ou negativo, o produto acabou
    else:
        return 'Produto esgotado'


# -------------------------------
# Função que exibe o relatório completo do produto
# -------------------------------
def exibir_relatorio(nome, inicial, vendas, final, situacao):
    print('\n RELATÓRIO DE ESTOQUE')  # Título do relatório
    print(f'Produto: {nome}')                # Mostra o nome do produto
    print(f'Estoque inicial: {inicial}')     # Mostra o estoque antes das vendas
    print(f'Unidades vendidas: {vendas}')    # Mostra quantas unidades foram vendidas
    print(f'Estoque final: {final}')         # Mostra o estoque restante
    print(f'Situação: {situacao}')           # Mostra a mensagem final (suficiente, baixo ou esgotado)


# ======================================
# PROGRAMA PRINCIPAL (executa tudo)
# ======================================

# Chama a função que pede o nome do produto e guarda na variável 'nome'
nome = pedir_nome_produto()

# Chama a função que pede o estoque inicial e guarda o valor em 'inicial'
inicial = pedir_estoque_inicial()

# Chama a função que pede a quantidade vendida e guarda em 'vendas'
vendas = pedir_vendas()

# Calcula o estoque final com base no inicial e nas vendas
final = calcular_estoque_final(inicial, vendas)

# Verifica a situação do estoque (suficiente, baixo ou esgotado)
situacao = verificar_situacao(final)

# Exibe o relatório completo com todos os dados
exibir_relatorio(nome, inicial, vendas, final, situacao)
```
